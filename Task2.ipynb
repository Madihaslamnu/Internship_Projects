```python
products = pd.read_csv('Olist_Data/olist_products_dataset.csv')
products[['product_id', 'product_category_name']].head()
```

```python
merged = items.merge(products[['product_id', 'product_category_name']], on='product_id', how='left')
merged = merged.merge(df[['order_id', 'order_purchase_timestamp']], on='order_id', how='left')
merged.head()
```

```python
merged['order_purchase_timestamp'] = pd.to_datetime(merged['order_purchase_timestamp'])
merged['month'] = merged['order_purchase_timestamp'].dt.to_period('M').astype(str)
merged[['order_purchase_timestamp', 'month']].head()
```

```python
top_categories = merged['product_category_name'].value_counts().head(5).index
subset = merged[merged['product_category_name'].isin(top_categories)]

wide = subset.pivot_table(
    index='month',
    columns='product_category_name',
    values='price',
    aggfunc='mean'
).round(2)

wide
```

```python
long_again = wide.reset_index().melt(
    id_vars='month',
    var_name='product_category_name',
    value_name='avg_price'
)

long_again.head(10)
```

```python
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])
```

```python
def compute_delay(row):
    return (row['order_delivered_customer_date'] - row['order_estimated_delivery_date']).days

df['delay_apply'] = df.apply(compute_delay, axis=1)
df[['order_delivered_customer_date', 'order_estimated_delivery_date', 'delay_apply']].head()
```

```python
df['delay_vectorized'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days
df[['delay_apply', 'delay_vectorized']].head()
```

```python
%timeit df.apply(compute_delay, axis=1)
```

```python
%timeit (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days
```
