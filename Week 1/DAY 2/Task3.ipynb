```python
import pandas as pd
import numpy as np

df={
    'customers': pd.read_csv('../../Olist_Data/olist_customers_dataset.csv'),
    'orders': pd.read_csv('../../Olist_Data/olist_orders_dataset.csv'),
    'order_items': pd.read_csv('../../Olist_Data/olist_order_items_dataset.csv'),
    'payments': pd.read_csv('../../Olist_Data/olist_order_payments_dataset.csv'),
    'reviews': pd.read_csv('../../Olist_Data/olist_order_reviews_dataset.csv'),
    'products': pd.read_csv('../../Olist_Data/olist_products_dataset.csv'),
    'sellers': pd.read_csv('../../Olist_Data/olist_sellers_dataset.csv'),
    'geolocation': pd.read_csv('../../Olist_Data/olist_geolocation_dataset.csv')
}
```

```python
df['order_items'][['price', 'freight_value']].describe()
```

```python

Q1 = df['order_items']['price'].quantile(0.25)
Q3 = df['order_items']['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"lower bound: {lower_bound}")
print(f"upper bound: {upper_bound}")

price_outliers_iqr = df['order_items'][(df['order_items']['price'] < lower_bound) | (df['order_items']['price'] > upper_bound)]
print(f"outliers flagged by IQR: {len(price_outliers_iqr)}")
```

```python
mean_price = df['order_items']['price'].mean()
std_price = df['order_items']['price'].std()

df['order_items']['price_zscore'] = (df['order_items']['price'] - mean_price) / std_price

price_outliers_zscore = df['order_items'][df['order_items']['price_zscore'].abs() > 3]
print(f"outliers flagged by z-score: {len(price_outliers_zscore)}")
```

```python
Q1_f = df['order_items']['freight_value'].quantile(0.25)
Q3_f = df['order_items']['freight_value'].quantile(0.75)
IQR_f = Q3_f - Q1_f

lower_bound_f = Q1_f - 1.5 * IQR_f
upper_bound_f = Q3_f + 1.5 * IQR_f

print(f"lower bound: {lower_bound_f}")
print(f"upper bound: {upper_bound_f}")

freight_outliers_iqr = df['order_items'][(df['order_items']['freight_value'] < lower_bound_f) | (df['order_items']['freight_value'] > upper_bound_f)]
print(f"outliers flagged by IQR: {len(freight_outliers_iqr)}")
```

```python
mean_freight = df['order_items']['freight_value'].mean()
std_freight = df['order_items']['freight_value'].std()

df['order_items']['freight_zscore'] = (df['order_items']['freight_value'] - mean_freight) / std_freight

freight_outliers_zscore = df['order_items'][df['order_items']['freight_zscore'].abs() > 3]
print(f"outliers flagged by z-score: {len(freight_outliers_zscore)}")
```
