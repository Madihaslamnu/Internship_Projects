```python
import os
import pandas as pd
os.listdir('Olist_Data')
```

```python
customers = pd.read_csv('Olist_Data/olist_customers_dataset.csv')
geolocation = pd.read_csv('Olist_Data/olist_geolocation_dataset.csv')
orders = pd.read_csv('Olist_Data/olist_orders_dataset.csv')
order_items = pd.read_csv('Olist_Data/olist_order_items_dataset.csv')
order_payments = pd.read_csv('Olist_Data/olist_order_payments_dataset.csv')
order_reviews = pd.read_csv('Olist_Data/olist_order_reviews_dataset.csv')
products = pd.read_csv('Olist_Data/olist_products_dataset.csv')
sellers = pd.read_csv('Olist_Data/olist_sellers_dataset.csv')
category_translation = pd.read_csv('Olist_Data/product_category_name_translation.csv')
```

```python
tables = {
    'customers': customers,
    'geolocation': geolocation,
    'orders': orders,
    'order_items': order_items,
    'order_payments': order_payments,
    'order_reviews': order_reviews,
    'products': products,
    'sellers': sellers,
    'category_translation': category_translation,
}

for name, t in tables.items():
    print(name, t.shape)
```

```python
orphans_items_in_orders = ~order_items['order_id'].isin(orders['order_id'])
print("order_items rows with no matching order:", orphans_items_in_orders.sum())
```

```python
orphans_orders_in_customers = ~orders['customer_id'].isin(customers['customer_id'])
print("orders rows with no matching customer:", orphans_orders_in_customers.sum())
```

```python
orphans_items_in_products = ~order_items['product_id'].isin(products['product_id'])
print("order_items rows with no matching product:", orphans_items_in_products.sum())
```

```python
orphans_items_in_sellers = ~order_items['seller_id'].isin(sellers['seller_id'])
print("order_items rows with no matching seller:", orphans_items_in_sellers.sum())
```

```python
orphans_payments_in_orders = ~order_payments['order_id'].isin(orders['order_id'])
print("order_payments rows with no matching order:", orphans_payments_in_orders.sum())
```

```python
orphans_reviews_in_orders = ~order_reviews['order_id'].isin(orders['order_id'])
print("order_reviews rows with no matching order:", orphans_reviews_in_orders.sum())
```

```python
orphans_products_in_translation = ~products['product_category_name'].isin(category_translation['product_category_name'])
print("products rows with no matching category translation:", orphans_products_in_translation.sum())
```
